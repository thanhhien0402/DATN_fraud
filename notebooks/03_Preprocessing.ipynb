```python
import pandas as pd
import numpy as np

# Đọc 3 tập dữ liệu đã chia ở Bước 12
train_df = pd.read_csv("../data/processed/train.csv")
val_df = pd.read_csv("../data/processed/validation.csv")
test_df = pd.read_csv("../data/processed/test.csv")

# Hiển thị kích thước dữ liệu
print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)

# Các cột không sử dụng trực tiếp
drop_columns = ["zipcodeOri", "zipMerchant", "customer"]

# Loại bỏ các cột trên
train_df = train_df.drop(columns=drop_columns)
val_df = val_df.drop(columns=drop_columns)
test_df = test_df.drop(columns=drop_columns)

# Kiểm tra kết quả
print("\nCác cột sau khi tiền xử lý:")
print(train_df.columns.tolist())

print("\nKích thước sau khi loại bỏ:")
print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

```python
# Kiểm tra các giá trị phân loại trong tập Train

categorical_columns = [
    "age",
    "gender",
    "merchant",
    "category"
]

for column in categorical_columns:

    print(f"\n===== {column} =====")

    print("Số lượng giá trị khác nhau:",
          train_df[column].nunique())

    print("Danh sách giá trị:")
    print(sorted(train_df[column].unique()))
```

```python
# Kiểm tra kiểu dữ liệu của các cột

print(train_df.dtypes)
```

```python
from sklearn.preprocessing import OneHotEncoder

# Danh sách các cột phân loại
categorical_columns = [
    "age",
    "gender",
    "merchant",
    "category"
]

# Khởi tạo bộ mã hóa
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# Học các nhóm giá trị từ tập Train
encoder.fit(train_df[categorical_columns])

print("Đã huấn luyện bộ mã hóa thành công!")
```

```python
# Lấy tên các cột sau khi mã hóa
encoded_columns = encoder.get_feature_names_out(
    categorical_columns
)

print("Tổng số cột sau mã hóa:",
      len(encoded_columns))

print("\nDanh sách các cột:")
print(encoded_columns)
```

```python
# Mã hóa các cột phân loại

X_train_cat = encoder.transform(
    train_df[categorical_columns]
)

X_val_cat = encoder.transform(
    val_df[categorical_columns]
)

X_test_cat = encoder.transform(
    test_df[categorical_columns]
)

print("Train:", X_train_cat.shape)
print("Validation:", X_val_cat.shape)
print("Test:", X_test_cat.shape)
```

```python
# Kiểm tra một số dòng dữ liệu sau mã hóa

encoded_train_df = pd.DataFrame(
    X_train_cat,
    columns=encoded_columns,
    index=train_df.index
)

print(encoded_train_df.head())
```

```python
# Lấy các đặc trưng số
X_train_num = train_df[["step", "amount"]].reset_index(drop=True)
X_val_num = val_df[["step", "amount"]].reset_index(drop=True)
X_test_num = test_df[["step", "amount"]].reset_index(drop=True)

# Chuyển dữ liệu One-Hot thành DataFrame
X_train_cat_df = pd.DataFrame(X_train_cat, columns=encoded_columns)
X_val_cat_df = pd.DataFrame(X_val_cat, columns=encoded_columns)
X_test_cat_df = pd.DataFrame(X_test_cat, columns=encoded_columns)

# Ghép đặc trưng số và đặc trưng phân loại
X_train = pd.concat([X_train_num, X_train_cat_df], axis=1)
X_val = pd.concat([X_val_num, X_val_cat_df], axis=1)
X_test = pd.concat([X_test_num, X_test_cat_df], axis=1)

# Lấy nhãn mục tiêu
y_train = train_df["fraud"].reset_index(drop=True)
y_val = val_df["fraud"].reset_index(drop=True)
y_test = test_df["fraud"].reset_index(drop=True)

# Kiểm tra kích thước
print("Train:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Test:", X_test.shape, y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd

# Khởi tạo bộ chuẩn hóa
scaler = StandardScaler()

# Tạo bản sao để không làm thay đổi dữ liệu gốc
X_train_scaled = X_train.copy()
X_val_scaled = X_val.copy()
X_test_scaled = X_test.copy()

# Biến đổi amount bằng log1p để giảm độ lệch phân phối
for X in [X_train_scaled, X_val_scaled, X_test_scaled]:
    X["amount"] = np.log1p(X["amount"])

# Chỉ huấn luyện scaler trên tập Train
numeric_columns = ["step", "amount"]

scaler.fit(X_train_scaled[numeric_columns])

# Áp dụng cùng scaler cho cả ba tập
X_train_scaled[numeric_columns] = scaler.transform(
    X_train_scaled[numeric_columns]
)

X_val_scaled[numeric_columns] = scaler.transform(
    X_val_scaled[numeric_columns]
)

X_test_scaled[numeric_columns] = scaler.transform(
    X_test_scaled[numeric_columns]
)

print("Train:")
print(X_train_scaled[numeric_columns].describe())

print("\nValidation:")
print(X_val_scaled[numeric_columns].describe())

print("\nTest:")
print(X_test_scaled[numeric_columns].describe())
```

```python
import os
import joblib

# Tạo thư mục lưu dữ liệu nếu chưa tồn tại
os.makedirs("../data/processed", exist_ok=True)

# Lưu đặc trưng đã chuẩn hóa
X_train_scaled.to_csv(
    "../data/processed/X_train_scaled.csv", index=False
)
X_val_scaled.to_csv(
    "../data/processed/X_val_scaled.csv", index=False
)
X_test_scaled.to_csv(
    "../data/processed/X_test_scaled.csv", index=False
)

# Lưu đặc trưng gốc (dành cho Random Forest và XGBoost)
X_train.to_csv("../data/processed/X_train.csv", index=False)
X_val.to_csv("../data/processed/X_val.csv", index=False)
X_test.to_csv("../data/processed/X_test.csv", index=False)

# Lưu nhãn
y_train.to_csv("../data/processed/y_train.csv", index=False)
y_val.to_csv("../data/processed/y_val.csv", index=False)
y_test.to_csv("../data/processed/y_test.csv", index=False)

# Lưu bộ mã hóa và bộ chuẩn hóa để tái sử dụng
joblib.dump(encoder, "../models/encoder.pkl")
joblib.dump(scaler, "../models/scaler.pkl")

print("Đã lưu thành công dữ liệu và bộ tiền xử lý!")
```
