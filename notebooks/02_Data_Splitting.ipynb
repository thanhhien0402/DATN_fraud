```python
import pandas as pd
import os

# Đọc dữ liệu gốc
df = pd.read_csv("../data/raw/bs140513_032310.csv")

# Kiểm tra dữ liệu
print("Kích thước dữ liệu:", df.shape)
print("Các cột:", df.columns.tolist())
```

```python
# Sắp xếp dữ liệu theo step tăng dần
df = df.sort_values(
    by="step",
    ascending=True
).reset_index(drop=True)

# Kiểm tra step nhỏ nhất và lớn nhất
print("Step nhỏ nhất:", df["step"].min())
print("Step lớn nhất:", df["step"].max())

# Kiểm tra dữ liệu đã sắp xếp
print("Dữ liệu đã sắp xếp:", df["step"].is_monotonic_increasing)
```

```python
# Chia dữ liệu theo step

train_df = df[
    (df["step"] >= 0) &
    (df["step"] <= 125)
].copy()

val_df = df[
    (df["step"] >= 126) &
    (df["step"] <= 152)
].copy()

test_df = df[
    (df["step"] >= 153) &
    (df["step"] <= 179)
].copy()
```

```python
print("===== KÍCH THƯỚC DỮ LIỆU =====")

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)

print("\nTổng số dòng:", 
      len(train_df) + len(val_df) + len(test_df))

print("Số dòng ban đầu:", len(df))
```

```python
print("===== KHOẢNG THỜI GIAN =====")

print(
    "Train:",
    train_df["step"].min(),
    "->",
    train_df["step"].max()
)

print(
    "Validation:",
    val_df["step"].min(),
    "->",
    val_df["step"].max()
)

print(
    "Test:",
    test_df["step"].min(),
    "->",
    test_df["step"].max()
)
```

```python
def check_fraud_distribution(data, name):

    print(f"\n===== {name} =====")

    print("Số giao dịch:", len(data))

    print("\nSố lượng từng nhãn:")
    print(data["fraud"].value_counts())

    print("\nTỷ lệ từng nhãn (%):")
    print(
        data["fraud"]
        .value_counts(normalize=True)
        .mul(100)
    )


check_fraud_distribution(train_df, "TRAIN")
check_fraud_distribution(val_df, "VALIDATION")
check_fraud_distribution(test_df, "TEST")
```

```python
# Kiểm tra các khoảng thời gian có bị giao nhau không

train_steps = set(train_df["step"].unique())
val_steps = set(val_df["step"].unique())
test_steps = set(test_df["step"].unique())

print("Train và Validation giao nhau:",
      len(train_steps & val_steps))

print("Train và Test giao nhau:",
      len(train_steps & test_steps))

print("Validation và Test giao nhau:",
      len(val_steps & test_steps))
```

```python
# Tạo thư mục processed nếu chưa tồn tại
os.makedirs("../data/processed", exist_ok=True)

# Lưu dữ liệu
train_df.to_csv(
    "../data/processed/train.csv",
    index=False
)

val_df.to_csv(
    "../data/processed/validation.csv",
    index=False
)

test_df.to_csv(
    "../data/processed/test.csv",
    index=False
)

print("Đã lưu thành công 3 tập dữ liệu!")
```
